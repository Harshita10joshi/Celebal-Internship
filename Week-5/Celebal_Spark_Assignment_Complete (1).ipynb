## Q1
**Limitations of MapReduce:** Writes intermediate data to disk, slow for iterative tasks, high latency, difficult programming model. **Why Spark:** In-memory processing, faster execution, supports SQL, ML, Streaming and Graph processing.

## Q2
Spark keeps intermediate data in RAM. Iterative ML algorithms reuse cached data instead of repeatedly reading from disk, making Spark significantly faster than disk-based systems like traditional MapReduce.

## Q3

```python
df.dropDuplicates(['user_id','transaction_date']).show()
```

## Q4

```python
from pyspark.sql.functions import avg
(df.filter(df.region=='West').groupBy('product_category').agg(avg('sale_amount').alias('Average_Sale'))).show()
```

## Q5
`.na.drop()` removes rows containing null values, whereas `.na.fill()` replaces null values with a specified value.

## Q5

```python
df.na.fill({'status':'Unknown'}).show()
```

## Q6

```python
from pyspark.sql.functions import count
(df.groupBy('city').agg(count('*').alias('cnt')).filter('cnt>100')).show()
```

## Q7
Spark DataFrames are immutable. Cleaning operations such as drop(), filter(), withColumnRenamed() and dropDuplicates() create a new DataFrame instead of modifying the original.

## Q9
Handling null values before sum() or avg() improves data quality, avoids incorrect calculations and ensures reliable analytical results.

## Q10

```python
from pyspark.sql.functions import col
from pyspark.sql.types import TimestampType
(df.withColumn('event_time',col('raw_timestamp').cast(TimestampType())).drop('raw_timestamp')).show()
```

## Q11
Shuffle redistributes data across partitions so records with the same key are grouped together. It is a wide transformation because data moves between partitions/executors, making it expensive.

## Q12

```python
from pyspark.sql.functions import col
(df.filter(col('email').isNotNull() & (col('username')!=''))).show()
```

## Q13

```python
from pyspark.sql.functions import min,max,mean
df.agg(min('price').alias('Min'),max('price').alias('Max'),mean('price').alias('Mean')).show()
```

## Q14
Using inferSchema=True on inconsistent date formats can infer wrong data types, convert dates to strings or nulls, and lead to incorrect filtering and aggregation. Defining schema explicitly is safer.

## Q15

```python
from pyspark.sql.functions import sum
(df.dropDuplicates().na.fill({'price':0}).groupBy('store_id').agg(sum('price').alias('Total_Revenue'))).show()
```
