# Spark Assignment
This notebook demonstrates Spark architecture concepts, schema handling, transformations, filtering, CSV vs Parquet, and performance insights.

```
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, upper
from pyspark.sql.types import *

spark=SparkSession.builder.appName("SparkAssignment").getOrCreate()

schema=StructType([
StructField("EmployeeID",IntegerType(),True),
StructField("Name",StringType(),True),
StructField("Department",StringType(),True),
StructField("Age",IntegerType(),True),
StructField("Salary",DoubleType(),True)
])

df=spark.read.csv("input/employees.csv",header=True,schema=schema)
df.show()
```

## Transformations
- Select columns
- Rename column
- Add uppercase department
- Filter salary > 50000
- Drop null values

```
df=df.select("EmployeeID","Name","Department","Salary")
df=df.withColumnRenamed("Salary","MonthlySalary")
df=df.withColumn("Department",upper(col("Department")))
filtered_df=df.filter(col("MonthlySalary")>50000).na.drop()
filtered_df.show()
```

```
filtered_df.write.mode("overwrite").option("header",True).csv("output_csv")
filtered_df.write.mode("overwrite").parquet("output_parquet")
print("Data written successfully.")
```

## Performance Insights
- Spark uses Lazy Evaluation.
- DAG optimizes execution before actions.
- Parquet is faster than CSV due to columnar storage and compression.
- Wide transformations (groupBy, join) cause shuffles.
- Predicate Pushdown improves Parquet read performance.
- Avoid collect() on large datasets; use show().
